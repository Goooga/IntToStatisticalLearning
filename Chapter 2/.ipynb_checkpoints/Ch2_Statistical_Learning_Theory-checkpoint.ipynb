### Задание 5.
Какими преимуществами обладают очень гибкие методы для регресси или классификации в сравнении с менее гибкими? При каких обстоятельствал более гибкие подходы могут быть предпочтительными по отношение к негибким? А когда стоит применять менее гибкие методы?

Менее гибкие:

+ (+) дают лучший результат с небольшим количеством наблюдений.
+ (+) модели хорошо подходят для статистических выводов.
+ (+) быстрее обучаются.
+ (+) устойчивы к переобучаются.
+ (-) плохо показывают себя, если в наблюдениях много нелинейных зависимостей

Гибкие:

+ (+) хорошо работают с нелинейными зависимостями в наблюдениях.
+ (-) подвержены переобучению.

### Задание 6.

Опишите отличия между параметрическим и непераметрическими методлами статистического обучения. Какими преимущетсвами обладают пераметрические методы применительно к задачам регресси или классификации в сравнении с непараметрическими? А какими недостатками?

> Параметрический подход упрощает задачу оценивания f(x), посколько гораздо легче оценить набор параметров в уже заданной модели, чем выполнять подгонку произвольной f(x) по данные. Возможным недостатком параметрического подхода является то, что выбранная нами модель обычно не будет соответствовать истинной не известной на форме f(x). Если выбранная нами модель окажется слишком далеко от истинной f(x), то наша оценка будет неэффективной. Можно попробовать решить это проблему с помощью выбора гибких моделей, однако это может превести к переобучению

> У непараметрических методо есть одно важное преимущество над непараметрическим: избегая предположения о функциональной форме f(x), мы имее возможность точно подгонять для нее множество разных форм. В тоже время у этих методов есть другой сещественный недостаток: в связи с тем, что они не сводят задачу оценивания f к определению небольшого количества параметров, для произведения точной оценки нам понадобиться очень большое количество наблюдений.

### Задание 7.

В таблице представлен обучающий набор данных, состоящий из шести наблюдений, трех предикторов и одного категориального отклика.

```python
import numpy as np 
import pandas as pd
```

```python
df = pd.DataFrame({'X1': [0, 2, 0, 0, -1, 1],
                   'X2': [3, 0, 1, 1, 0, -1],
                   'X3': [0, 0, 3, 2, 1, 1],
                   'Y': ['Red', 'Red', 'Red', 'Green', 'Green', 'Red']})

df
```

Представи, что мы используем этот набор данных для предсказания перменной Y при X1=X2=X3=0 и использованием метода k-ближайших соседей.

(a) Рассчитайте евклидово расстоние между каждым наблюдением и заданной контрольной точкой X1=X2=X3=0.

```python
X = df.drop(columns = 'Y', axis = 1)
dist_to_x = []
```

```python
X
```

```python
for i in X.index:
    dist_to_x.append(np.sqrt(sum(X.iloc[i].values**2)))
```

```python
for i in range(6):
    print(f'Расстоние от точки (0,0,0) до {i+1} наблюдения: {dist_to_x[i]}')
```

(b) Каким будет наше предсказание при K=1?

> Точка будет классифицирована как зеленая, т.к. первый ближайший сосед наблюдение 5(зеленая точка).

(c) Каким будет наше предсказание при K=3?

> Точка будет классифицирована как красная, т.к. из трех ближаших точек 2 красные и 1 зеленая.

(d) Если кривая байесовской решающей границы для нашего примера будет носить явно нелинейны характер, какое значение K лучше использовать: большое или маленькое?

> Лучши использовать маленькое значение k, т.к. так модель будет больше походить на байесовкую решающую границу.
