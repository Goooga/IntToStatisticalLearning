### Задание 3.

Рассмотрим индекс Джини, ошибку классафикаа и энтропию на примере простой классификации с двумя клкассами. Отобразите на графике кажду их этих велечини в зависимости от $\hat p_{m1}$. На оси ч далжны присутствовать $\hat p_{m1}$ в диапазоне от 0 до 1, а на оси y - значения каждого из показателелей.

```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
def E(p):
    return 1 - p

def Gini(p):
    return p*(1-p)

def Entropy(p):
    return -1*p*np.log(p)
```

```python
p = np.linspace(0, 1, 100)
```

```python
fig, ax = plt.subplots(figsize = (6,6))
ax.plot(p, [E(i) for i in p])
ax.plot(p, [Gini(i) for i in p])
ax.plot(p, [Entropy(i) for i in p])
ax.legend(['Classification error rate', 'Gine index', 'Entropy'])
ax.grid()
ax.set_title('Classification error rate, Gine index, Entropy')
ax.set_xlabel('p')
ax.set_ylabel('f(p)');
```
