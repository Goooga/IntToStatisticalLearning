## Задание 5.

Хорошо известно, что гребневая регрессия склонна давать похожие значения коэффициентов для коррелирующих перменных, тогда как метод лассо для таких переменных можеть давать сильно отличающиеся коэффициенты. Давайте исследуем это свойство на простом примере. Предположим, что n = 2, p = 2, $x_{11} = x_{12}, x_{21} = x_{22}$. Далее предположим, что $y_{1} + y_{2} = 0, x_{11} + x_{21} = 0$ и $x_{12} + x_{22} = 0, $ в результате чего оценка коэффициента свободного члена в моделях, подогнанных по методу наименьших квадратов, методу гребневой регрессии и методу лассо, будет равна нулю.

(a) Напишите для описанного случая уравнение оптимизации по методу гребневой регрессии.

Решение:

$e(\beta_{1}, \beta_{2}) = 
\displaystyle\sum_{i=1}^2 (y_{i} - \displaystyle\sum_{j=1}^2 \beta_{j}x_{ij})^{2} + \lambda \displaystyle\sum_{j=1}^p \beta_{j}^{2} = 
2(y_{1} - \beta_{1}x_{11}-\beta_{2}x_{11})^{2} + \lambda(\beta_{1}^{2}+\beta_{2}^{2})$ 

(b) Докажите, что в этом случае оценки коэффициентов гребневой регресси отвечают условию $\hat\beta_{1} = \hat\beta_{2}$.

Решение:

$e' = \left(\begin{array}{cc} 
-4x_{11}(y-\beta_{1}x_{11}-\beta_{2}x_{11})+2\lambda\beta_{1}\\
-4x_{11}(y-\beta_{1}x_{11}-\beta_{2}x_{11})+2\lambda\beta_{2}
\end{array}\right) =
\left(\begin{array}{cc}
0\\
0
\end{array}\right) \Rightarrow 
-4x_{11}(y-\beta_{1}x_{11}-\beta_{2}x_{11})+2\lambda\beta_{1} = -4x_{11}(y-\beta_{1}x_{11}-\beta_{2}x_{11})+2\lambda\beta_{2} \Rightarrow 
\beta_{1} = \beta_{2}$

(c) Напишите для этого случая уравнение оптимизации по методу лассо.

$e(\beta_{1}, \beta_{2}) = 
\displaystyle\sum_{i=1}^2 (y_{i} - \displaystyle\sum_{j=1}^2 \beta_{j}x_{ij})^{2} + \lambda \displaystyle\sum_{j=1}^p |\beta_{j}| = 
2(y_{1} - \beta_{1}x_{11}-\beta_{2}x_{11})^{2} + \lambda(|\beta_{1}|+|\beta_{2}|)$

(d) Покажите, что в этом сценарри оценки коэффициентов для лассо-модели $\hat\beta_{1}$ и $\hat\beta_{2}$ не уникальны. Иными словами, что существует множество возможных решений проблемы пункта (c).

Решение:

$sgn(x) = (|x|)' = \begin{equation}
\left\{\begin{split}
x = 1, x > 0 \\
x = 0, x = 0  \\
x = -1, x <0 
\end{split}\right.
 \end{equation}$

$e' = \left(\begin{array}{cc} 
-4x_{11}(y-\beta_{1}x_{11}-\beta_{2}x_{11})+2\lambda sgn(\beta_{1})\\
-4x_{11}(y-\beta_{1}x_{11}-\beta_{2}x_{11})+2\lambda sgn(\beta_{2})
\end{array}\right) =
\left(\begin{array}{cc}
0\\
0
\end{array}\right) \Rightarrow
sgn(\beta_{1}) = sgn(\beta_{2})$
