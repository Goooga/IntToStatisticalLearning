### Задание 1.

В этом упражнении мы поработаем с гиперплоскостями в двух измерениях.

(a) Изобразите гиперлоскость, соответствующую уравнению $ 1 + 3X_{1} - X_{2} = 0$. Обозначьте несколько точек на плоскости, для которых $1 + 3X_{1} - X_{2} > 0$ и несколько точек, для которых $1 + 3X_{1} - X_{2} < 0$.

(b) На том же графике изобразите гиперплоскость, соответствующую уравнению $-2 + X_{1} + 2X_{2} = 0$. Снова обозначьте несколько точек на плоскости, для которыз $-2 + X_{1} + 2X_{2} > 0$, и несколько точек, для которых $-2 + X_{1} + 2X_{2} < 0.$

```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
fig, ax = plt.subplots(figsize = (8, 6))

func1 = lambda x: 3*x + 1
func2 = lambda x: -x/2 + 1

func1X = [-5, 5]
func1Y = [func1(x) for x in func1X]

func2X = [-5, 5]
func2Y = [func2(x) for x in func2X]

pointX = [-1, -1, 0.6, 2]
pointY = [-1, 1, 1, 1.5]

ax.plot(func1X, func1Y, label = 'sf')
ax.plot(func2X, func2Y)
ax.scatter(pointX[:2], pointY[:2], marker = "o")
ax.scatter(pointX[2:], pointY[2:], marker = "v")
ax.legend(['1+3X_1 - X_2 = 0', '-2 + X_1 + 2X_2 = 0'])
ax.grid()
ax.set_xlabel('X1')
ax.set_ylabel('X2');
```

### Задание 2.

Мы видели, что при наличии p = 2 измерений линейная рещающая граница приобреате форму $\beta_{0} + \beta_{1}X_{1} + \beta_{2}X_{2} = 0$. Теперь исследуем нелинейную решающую границу.

(a) Изобразите краивую, соответствующую уравнению:

$(1 + X_{1})^{2} + (2-X_{2})^2 = 4$.

(b) На полученном графике обозначьте несколько точек на плоскости, для которых

$(1 + X_{1})^{2} + (2-X_{2})^2 > 4$.

и несколько для которыx

$(1 + X_{1})^{2} + (2-X_{2})^2 \leq 4$.

(c) Предположим, классификаток отност к синему классу наблюдения, для которых

$(1 + X_{1})^{2} + (2-X_{2})^2 > 4$.

К  каким классам будут отнесены наблюдения (0, 0), (-1, 1), (2, 2) и (3, 8)?

Синий: (0, 0), (2, 2), (3, 8)

Красный: (-1, 1)

```python
circle = plt.Circle((-1, 2), 2)

fig, ax = plt.subplots(figsize = (8, 6))

ax.add_patch(circle)
ax.legend(['(1 + X_{1})^{2} + (2-X_{2})^2 = 4'])
ax.set_xlim(-5, 5)
ax.set_ylim(-5, 5)
ax.scatter([-2,2], [-2,-2])
ax.scatter([-2,0], [2, 2], marker = '^')
ax.grid()
ax.set_xlabel('X1')
ax.set_ylabel('X2');
```

### Задание 3.

Теперь опробуем на небольшом игрушечном примере классификатор с максимальным зазором.

(a) У нас имеется n = 7 наблюдений и p = 2размерности. Для каждого наблюдщения проставлен свой класс, как показаоно в табл 9.1. Изобразите эта наблюдения на графике.

```python
redX = [3, 2, 4, 1]
redY = [4, 2, 4, 4]

blueX = [2, 4, 4]
blueY = [1, 3, 1]

fig, ax = plt.subplots(figsize = (8, 6))

ax.scatter(redX, redY, color = 'red')
ax.scatter(blueX, blueY, color = 'blue')
```

(b) Нарисуйте оптимальную разделяющую гиперплоскость и приведите ее уравнение вида(9.1)

```python
redX = [3, 2, 4, 1]
redY = [4, 2, 4, 4]

blueX = [2, 4, 4]
blueY = [1, 3, 1]

fig, ax = plt.subplots(figsize = (8, 6))

ax.scatter(redX, redY, color = 'red')
ax.scatter(blueX, blueY, color = 'blue')

funcDirBlue = lambda x: x - 1 
funcDirRed = lambda x: x
svmFunc = lambda x: x - 1/2

X = [0, 5]
ax.plot(X, [funcDirBlue(x) for x in X], linestyle = ':', color = 'red')
ax.plot(X, [funcDirRed(x) for x in X], linestyle = ':', color = 'red')
ax.plot(X, [svmFunc(x) for x in X])
ax.set_xlabel('X1')
ax.set_ylabel('X2')
ax.grid()
```
