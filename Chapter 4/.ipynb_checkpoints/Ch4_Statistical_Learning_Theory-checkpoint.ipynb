$p(X) = \frac{e^{\beta_0 + \beta_1 X}}{1 + e^{\beta_0 + \beta_1 X}} =  \frac{e^{\beta_0 + \beta_1 X} + 1 - 1}{1 + e^{\beta_0 + \beta_1 X}} = 1 - \frac{1}{1 + e^{\beta_0 + \beta_1 X}} \Rightarrow 1 + e^{\beta_0 + \beta_1 X} = \frac{1}{1 - p(X)}$ 

$p(X)({1 + e^{\beta_0 + \beta_1 X}}) =  e^{\beta_0 + \beta_1 X} \Rightarrow \frac{p(X)}{1-p(X)} = e^{\beta_0 + \beta_1 X}$

## Задание 6.

Представьте, что мы собрали данные для группы студентов курса статистики с двемя предикторами $X_{1}$ = часы обучения по предмету и $X_{2}$ = средний балл в школе, а также откликом Y = индикатор получения высшено балла. Мы выполнили подгонку лигистической регресси и получили оценки коэффициентов $\hat \beta_{0} = -6$, $\hat \beta_{1} = 0.05$, $\hat \beta_{2} = 1$.

(a) Оцените вероятсноть того, что студент со средним баллом в школе 3.5, обучавшийся статистике 40 часов, получит на экзамене высший балл.


```python
import numpy as np
```

```python
f = lambda x1, x2: np.exp(-6 + 0.05 * x1 + 1*x2)/(1 + np.exp(-6 + 0.05 * x1 + 1*x2))
```

```python
print(f'Вероятность получить высший балл равна: {f(40, 3.5)}')
```

(b) Сколько студенту (a) необходимо часов обучения, чтобы вероятсноть получения им высшего балла на экзамене составила 50%?

> Ответ: 50 часов.
