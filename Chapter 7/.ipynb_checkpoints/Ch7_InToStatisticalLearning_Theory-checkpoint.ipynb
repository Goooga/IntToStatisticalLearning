## Теоритические задания.

В этой главе упоминалось, что кубической регрессионный сплайн с одним уздлм в точке $\xi$ можно получить с помощью базисных функции вида $x, x^2, x^3, (x-\xi)^3_{+}, где (x_\xi)^3_{+}$, если $x > \xi$, и равна 0 в противном случае. Теперь покажем, что функция вида

$f(x) = \beta_{0} + \beta_{1}x + \beta_{2}x^2 + \beta_{3}x^3 + \beta_{4}(x-\xi)^3_{+}$

дествительно является кубическим регриссионным сплайном, вне зависимости от значения $\beta_{0}, \beta_{1}, \beta_{2}, \beta_{3}, \beta_{4}.$

(a) Найдите такой кубический полином 

$f_{1}(x) = a_{1} + b_{1}x + c_{1}x^2 + d_{1}x^3,$

у которого $f(x) = f_{1}(x)$ для всех $x \leq \xi$. Выразите $a_{1}, b_{1}, c_{1}, d_{1}$ в виде $\beta_{0}, \beta_{1}, \beta_{2}, \beta_{3}, \beta_{4}.$

Решение:

$f(x) = f_{1}(x), x \leq \xi$

$\beta_{0} + \beta_{1}x + \beta_{2}x^2 + \beta_{3}x^3 + 0 = a_{1} + b_{1}x + c_{1}x^2 + d_{1}x^3 \Rightarrow \beta_{0} = a_{1}, \beta_{1} = b_{1}, \beta_{2} = c_{1}, \beta_{3} = d_{1},$

$f_{1}(x) = \beta_{0} + \beta_{1}x + \beta_{2}x^2 + \beta_{3}x^3$

(b) Найдите такой кубический полином 

$f_{2}(x) = a_{2} + b_{2}x + c_{2}x^2 + d_{2}x^3,$

у которого $f(x) = f_{2}(x)$ для всех $x > \xi$. Выразите $a_{2}, b_{2}, c_{2}, d_{2}$ в виде $\beta_{0}, \beta_{1}, \beta_{2}, \beta_{3}, \beta_{4}.$

Решение:

$f(x) = \beta_{0} + \beta_{1}x + \beta_{2}x^2 + \beta_{3}x^3 + \beta_{4}(x-\xi)^3_{+} = (\beta_{0} + \beta_{4}\xi^3) + (\beta_{1} + 2\xi^2\beta_{4})x + (\beta_{2} + 2\xi\beta_{4})x^2 + (\beta_{3} + \beta_{4})x^3, x > \xi $

$f(x) = f_{2}(x)$

$(\beta_{0} + \beta_{4}\xi^3) + (\beta_{1} + 2\xi^2\beta_{4})x + (\beta_{2} + 2\xi\beta_{4})x^2 + (\beta_{3} + \beta_{4})x^3 = a_{2} + b_{2}x + c_{2}x^2 + d_{2}x^3 \Rightarrow (\beta_{0} + \beta_{4}\xi^3) = a_{2}, (\beta_{1} + 2\xi^2\beta_{4}) = b_{2}, (\beta_{2} + 2\xi\beta_{4}) = c_{2}, (\beta_{3} + \beta_{4}) = d_{2}$

$f_{2}(x) = (\beta_{0} + \beta_{4}\xi^3) + (\beta_{1} + 2\xi^2\beta_{4})x + (\beta_{2} + 2\xi\beta_{4})x^2 + (\beta_{3} + \beta_{4})x^3$

(c) Покажите, что $f(x)$ непрерывна в точке $\xi$.

Решение:

1) $f(x)$ определена в $\xi$:
   
   $f(\xi) = \beta_{0} + \beta_{1}x + \beta_{2}x^2 + \beta_{3}x^3$

2) $\lim_{x\rightarrow \xi_{-0}}f(x) = \lim_{x\rightarrow \xi_{+0}}f(x)$

   $\lim_{x\rightarrow \xi_{-0}}f(x) = \beta_{0} + \beta_{1}x + \beta_{2}x^2 + \beta_{3}x^3$

   $\lim_{x\rightarrow \xi_{+0}}f(x) = \beta_{0} + \beta_{1}x + \beta_{2}x^2 + \beta_{3}x^3$

(d) Покажите, что $f^{'}(x)$ неперывна в точке $\xi$.

Решение:

1) $f^{'}(x)$ определена $\xi$:
   
   $f(\xi) = \beta_{1} + 2\beta_{2}x + 3\beta_{3}x^2$

2) $\lim_{x\rightarrow \xi_{-0}}f^{'}(x) = \lim_{x\rightarrow \xi_{+0}}f^{'}(x)$

   $\lim_{x\rightarrow \xi_{-0}}f^{'}(x) = \beta_{1} + 2\beta_{2}x + 3\beta_{3}x^2$
   
   $\lim_{x\rightarrow \xi_{+0}}f^{'}(x) = \beta_{1} + 2\beta_{2}x + 3\beta_{3}x^2$

(e) Покажите, что $f^{''}(x)$ неперывна в точке $\xi$.

Решение:

1) $f^{''}(x)$ определена $\xi$:
   
   $f(\xi) = 2\beta_{2} + 6\beta_{3}x$

2) $\lim_{x\rightarrow \xi_{-0}}f^{''}(x) = \lim_{x\rightarrow \xi_{+0}}f^{''}(x)$

   $\lim_{x\rightarrow \xi_{-0}}f^{''}(x) = 2\beta_{2} + 6\beta_{3}x$
   
   $\lim_{x\rightarrow \xi_{+0}}f^{''}(x) = 2\beta_{2} + 6\beta_{3}x$

### Задание 2.

Допустим, кривая $\hat g$ вычислена для гладкой аппроксимации некоторого набора из n точек по следующей формуле:

$\hat g = argmin_{g}(\sum_{i=1}^n (y_{i} - g(x_{i}))^2 + \lambda \int[g^{(m)}(x)]^2dx),$

где $g^{m}$ представляет m-ю производную g. Предложите эскиз кривой $\hat g$ для каждого из следующих сценариев.

(a) $\lambda = \infty, m = 0$

```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
x = [i for i in range(10)]
y = [0]*10

fig, ax = plt.subplots(figsize = (6,6))
ax.plot(x,y)
ax.grid()
ax.set_xlabel('x')
ax.set_ylabel('y');
```

(b) $\lambda = \infty, m = 1$

```python
x = [i for i in range(10)]
y = x

fig, ax = plt.subplots(figsize = (6,4))
ax.plot(x,y)
ax.set_xlabel('x')
ax.set_ylabel('y');
```

(c) $\lambda = \infty, m = 2$

```python
x = [i for i in range(10)]
y = [(i-4)**2 for i in x]

fig, ax = plt.subplots(figsize = (6,4))
ax.plot(x,y)
ax.grid()
ax.set_xlabel('x')
ax.set_ylabel('y');
```

(d) $\lambda = \infty, m = 3$

```python
x = [i for i in range(10)]
f = lambda x: (x-4)**3
y = [f(i) for i in x]

fig, ax = plt.subplots(figsize = (6,4))
ax.plot(x,y)
ax.grid()
ax.set_xlabel('x')
ax.set_ylabel('y');
```

(e) $\lambda = 0, m = 3$

```python
x = [i for i in range(10)]
y = [i+np.random.randint(0,5) for i in x]

fig, ax = plt.subplots(figsize = (6,4))
ax.plot(x,y)
ax.grid()
ax.scatter(x,y)
ax.set_xlabel('x')
ax.set_ylabel('y');
```

### Задание 3.

Препдположим, что мы выполняем подгонку кривой с базисным функциями $b_{1}(X) = X, b_{2}(X) = (X-1)^2I(X\geq 1)$ Подгоняем линейную модель:

$Y = \beta_{0}+\beta_{1}b_{1}(X)+\beta_{2}b_{2}(X)+\varepsilon$

и получаем оченки коэффициентов $\hat\beta_{0}=1, \hat\beta_{1}=1, \hat\beta_{2}=-2$. Нарисуйте эскиз оцененной кривой меджу точками X = -2 и X = 2. 

```python
def b2(x):
    if x>=1:
        return (x-1)**2
    return 0

X = np.linspace(-2, 2, 100)
y = [1 + x - 2*b2(x) for x in X]

fig, ax = plt.subplots(figsize = (6,4))
ax.plot(X,y)
ax.grid()
ax.set_xlabel('x')
ax.set_ylabel('y');
```

### Задание 4.

Препдположим, что мы выполняем подгонку кривой с базисным функциями $b_{1}(X) = I(0\leq X\leq2)-(X-1)I(1\leq X\leq2), b_{2}(X) = (X-3)I(3\leq X\leq4)+I(4<X\leq5)$ Подгоняем линейную модель:

$Y = \beta_{0}+\beta_{1}b_{1}(X)+\beta_{2}b_{2}(X)+\varepsilon$

и получаем оченки коэффициентов $\hat\beta_{0}=1, \hat\beta_{1}=1, \hat\beta_{2}=3$. Нарисуйте эскиз оцененной кривой меджу точками X = -2 и X = 6. 

```python
def b1(x):
    i1 = 0
    i2 = 0
    if 0<=x<=2:
        i1= 1
    if 1<=x<=2:
        i2 = 1
    return i1 - (x-1)*i2

def b2(x):
    i1 = 0
    i2 = 0
    if 3<=x<=4:
        i1 = 1
    if 4<x<=5:
        i2 = 1
    return (x-3)*i1 + i2

X = np.linspace(-2, 6, 100)
y = [1 + b1(x) + 3*b2(x) for x in X]
fig, ax = plt.subplots(figsize = (6,4))
ax.plot(X, y)
ax.grid()
ax.set_xlabel('x')
ax.set_ylabel('y')
```

### Задание 5.

Рассмотрим две кривые $\hat g_{1}$ и $\hat g_{2}$ вида:

$$\hat g_{1} = argmin_{g} (\sum_{i=1}^n (y_{i} - g(x_{i}))^2 + \lambda \int[g^{(3)}(x)]^2dx),$$

$$\hat g_{2} = argmin_{g}(\sum_{i=1}^n (y_{i} - g(x_{i}))^2 + \lambda \int[g^{(4)}(x)]^2dx),$$

(a) При $\lambda \rightarrow \infty$ у какой из этих кривых будет меньше RSS на обучающих данных?

Ответ: $\hat g_{2}$

(b) При $\lambda \rightarrow \infty$ у какой из этих кривых будет меньше RSS на контрольных данных?

Ответ: если данные описываются полиномом 1-ой или 2-оЙ степени, то $\hat g_{1}$, иначе $\hat g_{2}$

(c) При $\lambda = 0$ у какой из этих кривых будет меньше RSS на обучающих и контрольных данных?

Ответ: $\hat g_{1}$ и $\hat g_{2}$ будут интерполировать обучающие данные и покажут одинаковый результат на обучающих и на проверочных данных.
